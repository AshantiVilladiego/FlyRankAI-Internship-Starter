## 3. Success metric

The primary success metric will be **Precision@K**, with an initial focus on **Precision@50**.

Precision@50 measures how many of the 50 highest-ranked pages correspond to the target outcome or decline proxy. This is appropriate because the real-world action is to give a limited number of pages to a human content team for review.

A useful ranking system should therefore place the strongest opportunities near the top of the queue rather than only achieving good overall prediction performance.

I will also consider ranking metrics such as Average Precision when evaluating the model, but Precision@50 is especially useful for connecting model performance to a realistic review capacity.

### Unit of analysis

The page is the appropriate unit of analysis because the real-world decision is also made at the page level: a content reviewer needs to determine which individual pages should be prioritized for review.

## 5. Why ML beats a fixed rule here

A fixed rule could rank pages using simple thresholds, such as selecting pages with declining traffic or pages older than a certain number of days. This would be easy to understand, but it may miss interactions between multiple signals.

ML may be useful because content opportunity can depend on combinations of factors such as impressions, clicks, sessions, content age, trend direction, CTR, and search position. A model can learn patterns across these variables and produce a more flexible ranking than a single manually chosen threshold.

However, ML is only worthwhile if it improves the ranking of useful pages compared with a transparent baseline. The goal is therefore not to use ML simply because it is more sophisticated, but to test whether it provides better decision support than a reasonable fixed-rule approach.
