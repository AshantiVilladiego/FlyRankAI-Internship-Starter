## 3. Success metric

The primary success metric will be **Precision@K**, with an initial focus on **Precision@50**.

Precision@50 measures how many of the 50 highest-ranked pages correspond to the target outcome or decline proxy. This is appropriate because the real-world action is to give a limited number of pages to a human content team for review.

A useful ranking system should therefore place the strongest opportunities near the top of the queue rather than only achieving good overall prediction performance.

I will also consider ranking metrics such as Average Precision when evaluating the model, but Precision@50 is especially useful for connecting model performance to a realistic review capacity.

## 5. Why ML beats a fixed rule here

A fixed rule could prioritize pages using a small number of manually selected thresholds, such as traffic decline or content age. This would be transparent and may be a useful baseline.

However, the problem involves multiple observable signals that may interact in ways that are difficult to express with a single hand-written rule. ML is worth testing if it can use these signals to produce a better-ranked queue than the transparent baseline.

The goal is therefore not to assume that ML is automatically better. I will compare the learned approach against a fixed-rule baseline using Precision@K, with Precision@50 providing a direct measure of how useful the highest-priority pages are for a limited human review capacity.
